```python
import pandas as pd
df = pd.read_csv("songs.csv")
df["mood"] = df["mood"].str.lower()
df["genre"] = df["genre"].str.lower()

print("\nSONG RECOMMENDATION SYSTEM READY ✔\n")
user_mood = input("Enter your mood (happy / sad / chill / romantic / energetic / confident): ").strip().lower()
user_genre = input("Enter preferred genre (pop / indie / rock / jazz / hip-hop / alternative / psych rock): ").strip().lower()
results = df[(df["mood"] == user_mood) & (df["genre"] == user_genre)]
if results.empty:
    print("\nNo exact match found for both mood and genre.")
    print("Showing songs matching only your mood:\n")
    results = df[df["mood"] == user_mood]
if results.empty:
    print("\nNo songs fit your mood. Showing popular songs instead:\n")
    results = df.sample(5)
print("🎵 Recommended Songs 🎵\n")
for index, row in results.iterrows():
    print(f"- {row['song']} by {row['artist']}  ({row['genre'].title()}, {row['mood']})")
```
